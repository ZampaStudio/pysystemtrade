```python
import pandas as pd
import time
from systems.forecasting import TradingRule, Rules
from systems.provided.rules.ewmac import ewmac_forecast_with_defaults as ewmac
from sysdata.config.configdata import Config
from systems.forecast_combine import ForecastCombine
from systems.basesystem import System
from systems.accounts.accounts_stage import Account
from systems.forecast_scale_cap import ForecastScaleCap
from systems.rawdata import RawData
from systems.positionsizing import PositionSizing
from sysdata.sim.csv_futures_sim_data import csvFuturesSimData

# my_account = Account()
# scale_and_cap = ForecastScaleCap()
# combiner = ForecastCombine()
# raw_data = RawData()
# position_size = PositionSizing()
# data = csvFuturesSimData()

# instruments
portfolio_intruments =  [
     "NASDAQ",
     "SP500",
     "US10",
     "RUSSELL",
     "EUR",
     "GOLD",
     "SILVER",
     "EUROSTX",
     "BITCOIN",
     "CRUDE_W_mini",
     "LIVECOW"
     ]

test_instrument_code = "NASDAQ"
main_rule_name = "ewmac_32_128"

# define rules
ewmac_5_20 = TradingRule((ewmac, [], dict(Lfast=5, Lslow=20)))
ewmac_20_60 = TradingRule((ewmac, [], dict(Lfast=20, Lslow=60)))
ewmac_32_128 = TradingRule((ewmac, [], dict(Lfast=32, Lslow=128)))
ewmac_60_150 = TradingRule((ewmac, [], dict(Lfast=60, Lslow=150)))
ewmac_100_200 = TradingRule((ewmac, [], dict(Lfast=100, Lslow=200)))

system_rules = Rules(dict(ewmac_5_20=ewmac_5_20, 
                      ewmac_20_60=ewmac_20_60,
                      ewmac_32_128=ewmac_32_128,
                      ewmac_60_150=ewmac_60_150,
                      ewmac_100_200=ewmac_100_200))
```

```python
# Forecast scalar (fixed)
# https://github.com/robcarver17/pysystemtrade/blob/develop/docs/backtesting.md#forecast-scalar-fixed

def test_forecast_scalar_fixed():
    config = Config()
    config.use_forecast_scale_estimates = False
    
    data = csvFuturesSimData()
    scale_and_cap = ForecastScaleCap()
    system = System([scale_and_cap, system_rules], data, config)
    
    raw_forecasts = system_rules.get_raw_forecast(test_instrument_code, "ewmac_32_128") #system.rules.get_raw_forecast(test_instrument_code, "ewmac_32_128")
    scalar = scale_and_cap.get_forecast_scalar(test_instrument_code, "ewmac_32_128")
    scaled_forecasts = scale_and_cap.get_scaled_forecast(test_instrument_code, "ewmac_32_128")
    capped_forecasts = scale_and_cap.get_capped_forecast(test_instrument_code, "ewmac_32_128")
    
    return pd.DataFrame({
        'raw': raw_forecasts,
        'scalar': scalar,
        'scaled': scaled_forecasts,
        'capped': capped_forecasts
    })

test_forecast_scalar_fixed().tail(10)
```

```python
# Forecast scalar (estimated)
# https://github.com/robcarver17/pysystemtrade/blob/develop/docs/backtesting.md#forecast-scalar-estimated

# Nota: per qualche motivo il risultato in c# è leggermente diverso da quello di pysystemtrade a causa di questa funzione:
# avg_abs_value = x.rolling(window=window, min_periods=min_periods).mean()

# L'input di questa funzione sembra corrispondere, a parte per alcuni NaN iniziali che in c# sono già stati rimossi
# Credo sia dovuto all'approssimazione ma non sono convinto.

def test_forecast_scalar_estimated():
    config = Config()
    config.use_forecast_scale_estimates=True
    config.forecast_scalar_estimate=dict(pool_instruments=False, func="sysquant.estimators.forecast_scalar.forecast_scalar", window=250000, min_periods=500, backfill=True)
    
    data = csvFuturesSimData()
    scale_and_cap = ForecastScaleCap()
    system = System([scale_and_cap, system_rules], data, config)
    
    raw_forecasts = system_rules.get_raw_forecast(test_instrument_code, "ewmac_32_128") #system.rules.get_raw_forecast(test_instrument_code, "ewmac_32_128")
    scalar = scale_and_cap.get_forecast_scalar(test_instrument_code, "ewmac_32_128")
    scaled_forecasts = scale_and_cap.get_scaled_forecast(test_instrument_code, "ewmac_32_128")
    capped_forecasts = scale_and_cap.get_capped_forecast(test_instrument_code, "ewmac_32_128")
    
    return pd.DataFrame({
        'raw': raw_forecasts,
        'scalar': scalar,
        'scaled': scaled_forecasts,
        'capped': capped_forecasts
    })

start = time.perf_counter()
df_pooling = test_forecast_scalar_estimated().tail(10)
elapsed = time.perf_counter() - start
print(f"Loaded in: {elapsed}")
df_pooling
```

```python
# Forecast weights (fixed)
# https://github.com/robcarver17/pysystemtrade/blob/develop/docs/backtesting.md#forecast-weights-fixed

def test_forecast_combination_fixed():
    config = Config()
    config.instruments = portfolio_intruments
    config.use_forecast_weight_estimates=False
    #config.forecast_weights=dict(ewmac_20_60=0.5, ewmac_32_128=0.5)
    
    data = csvFuturesSimData()
    scale_and_cap = ForecastScaleCap()
    combine= ForecastCombine()
    system = System([scale_and_cap, combine, system_rules], data, config)

    capped_forecasts_1 = scale_and_cap.get_capped_forecast(test_instrument_code, "ewmac_5_20")
    capped_forecasts_2 = scale_and_cap.get_capped_forecast(test_instrument_code, "ewmac_20_60")
    capped_forecasts_3 = scale_and_cap.get_capped_forecast(test_instrument_code, "ewmac_32_128")
    capped_forecasts_4 = scale_and_cap.get_capped_forecast(test_instrument_code, "ewmac_60_150")
    capped_forecasts_5 = scale_and_cap.get_capped_forecast(test_instrument_code, "ewmac_100_200")
    combined_forecasts = combine.get_raw_combined_forecast_before_mapping(test_instrument_code)

    return pd.DataFrame({
        'ewmac_5_20': capped_forecasts_1,
        'ewmac_20_60': capped_forecasts_2,
        'ewmac_32_128': capped_forecasts_3,
        'ewmac_60_150': capped_forecasts_4,
        'ewmac_100_200': capped_forecasts_5,
        'combined (weighted)': combined_forecasts
    })

start = time.perf_counter()
df_pooling = test_forecast_combination_fixed().tail(10)
elapsed = time.perf_counter() - start
print(f"Loaded in: {elapsed}")
df_pooling
```
